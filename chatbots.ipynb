```python
import os
from dotenv import load_dotenv
load_dotenv()


groq_api_key = os.getenv("GROQ_API_KEY")
```

```python
from langchain_groq import ChatGroq
model = ChatGroq(model = "llama-3.3-70b-versatile",groq_api_key = groq_api_key)
model
```

```python
from langchain_core.messages import HumanMessage
model.invoke([HumanMessage(content = "Hi, My name is pavan I am learning GEN AI")])
```

```python
#Message History.
"""
We can use a Message History Class to wrap our model and make it stateful.
This will keep track of inputs and outpits of the models and store it in the some data store.
Future interactions will then load those messages and pass them into chain as part of the input.

"""
```

```python
from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.chat_history import BaseChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory

store = {}
def get_session_history(session_id:str)->BaseChatMessageHistory:
    if session_id not in store:
        store[session_id] = ChatMessageHistory()
    return store[session_id]


with_message_history = RunnableWithMessageHistory(model, get_session_history)
```

```python
config = {"configurable":{"session_id":"chat1"}}
```

```python
response=with_message_history.invoke(
    [
        HumanMessage(content = "Hi My name is pavan I am learning GEN AI")
    ],
    config = config
)
```

```python
response.content
```

```python
#Prompt Template.
"""
Prompt template help to turn raw user info into a format that the LLM can work.
In this case, the user just a message , which are passing to the LLM.

"""
```

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
prompt = ChatPromptTemplate.from_messages(
    [
        ("system","you are a helpful assistant. Answer all the questions to best of your ability"),
        MessagesPlaceholder(variable_name="messages")
    ]

)
chain = prompt | model
```

```python
chain.invoke({"messages":[HumanMessage(content = "Hi My name is pavan")]})
```

```python
with_message_history = RunnableWithMessageHistory(chain,get_session_history)
```

```python
config = {"configurable":{"session_id":"chat3"}}
response = with_message_history.invoke(
    [
        HumanMessage(content = "Hi My name is Pavan")],
        config = config,
    
)
response.content
```

```python
prompt = ChatPromptTemplate.from_messages(
    [
        (
            "system",
            "you are a helpful assistant. Answer all the questions to the best of your ability in {language}.",
        ),
        MessagesPlaceholder(variable_name="messages")
    ]

)
chain = prompt | model
```

```python
response = chain.invoke({"messages":[HumanMessage(content = "Hi my name is pavan")],"language":"Hindi"})
response.content
```

```python
with_message_history = RunnableWithMessageHistory(
    chain,
    get_session_history,
    input_messages_key="messages"
)
```

```python
config = {"configurable":{"session_id":"chat4"}}
response = with_message_history.invoke(

{
    "messages": [HumanMessage(content = "Hi, I am Pavan")],"language":"hindi"
},
config = config
)
response.content
```

```python
#Managing the chat history.
"""
Using session id we have to been able to track the chat history.
This chat history can grow in size and we have to learn how to manage the chat history.
without managing the chat history , the list of messages will grow unbound and potentially overflow the context window of LLM.
Hence, we have add applications to limit the size of teh messages.

"trim_messages" helps to reduce the number of messages we want to send to the LLM along with specifying the number of tokens we want to restrict to.
To use the install transformers module to use the hugging face.

"""
```

```python
from langchain_core.messages import SystemMessage, trim_messages
trimmer = trim_messages(
    max_tokens=200,
    strategy ="last",
    token_counter=model,
    include_system=True,
    allow_partial=False,
    start_on="human"

)

messages = [
    SystemMessage(content = "You are good assistent. Answer with your ability"),
    HumanMessage(content ="Hi, I am Pavan" ),
    AIMessage(content = "Hi! Nice meeting you pavan"),
    HumanMessage(content = "I like to play Kabaddi"),
    AIMessage(content = "That's a wonderfull game"),
    HumanMessage(content = "How many members are going to play kabaddi"),
    AIMessage(content = "In each team 7 members"),
    HumanMessage(content = "How can we survive in kabaddi game"),
    AIMessage(content = "Try to do daily practice "),
    HumanMessage(content = "Thank you"),
    AIMessage(content = "Yes ofcourse ! Thank you Pavan, Nice talking with you")
]
trimmer.invoke(messages)
```

```python
from operator import itemgetter
from langchain_core.runnables import RunnablePassthrough
from langchain_core.messages import HumanMessage

chain = (
    RunnablePassthrough.assign(
        messages=itemgetter("messages") | trimmer
    )
    | prompt
    | model
)

response = chain.invoke(
    {
        "messages": messages + [
            HumanMessage(content="Which game do I like to play?")
        ],
        "language": "English",
    }
)

response.content
```

```python
#Lets wrap with this message history.
with_message_history = RunnableWithMessageHistory(
    chain,
    get_session_history,
    input_messages_key="messages",
)
config = {"configurable":{"session_id":"chat5"}}
```

```python
response = with_message_history.invoke(
    {
        "messages":messages + [HumanMessage(content = "whats my name")],
        "language":"English",

    },
    config = config,
)
response.content
```
